> We have a knapsack of capacity weight C (a positive integer) and n types of objects.
Each object of the ith type has weight wi and profit pi (all wi and all pi are positive
integers, i = 0, 1, …, n-1). There are unlimited supplies of each type of objects. Find
the largest total profit of any set of the objects that fits in the knapsack.
Let P(C) be the maximum profit that can be made by packing objects into the knapsack
of capacity C.


 #### (Q1) Give a recursive definition of the function P(C).

### 📌 Recursive Definition of P(C) — Unbounded Knapsack

Let `P(C)` be the maximum profit for a knapsack with capacity `C`, given `n` types of items.  
Each item type `i` has weight `w[i]` and profit `p[i]`.  
You can take an unlimited number of each item.

We define `P(C)` as:

\[
P(C) = \begin{cases}
0 & \text{if } C = 0 \\
\max\limits_{\substack{0 \leq i < n \\ w_i \leq C}} \left( p_i + P(C - w_i) \right) & \text{if } C > 0
\end{cases}
\]

**Explanation:**
- Base case: If the knapsack is empty, profit is 0.
- Recursive case: For every item that can fit, try taking it and add its profit to the max profit of the remaining capacity. Choose the maximum of these options.

#### (Q3) Give a dynamic programming algorithm to compute the maximum profit, given a knapsack of capacity C, n types of objects with weights wi and profits pi using the bottom up approach.

Function KNAPSACK(C, weights[0...n-1], profits[0...n-1]):

    # Initialize a DP array of size C + 1
    Create array dp[0...C] and set all values to 0
    # dp[c] will store the maximum profit for capacity c

    # Loop through each capacity from 1 to C
    For c from 1 to C:
        # Try every item type
        For i from 0 to n - 1:
            If weights[i] ≤ c then:
                # Try including the item
                dp[c] ← max(dp[c], dp[c - weights[i]] + profits[i])

    # Return the maximum profit for capacity C
    Return dp[C]


#### (Q4) Code your algorithm in a programming language

Algorithm code:

```python
def knapsack(C, weights, profits):
    dp = [0] * (C + 1)
    for c in range(1, C + 1):
        for i in range(len(weights)):
            if weights[i] <= c:
                dp[c] = max(dp[c], dp[c - weights[i]] + profits[i])
    return dp[C], dp
```

Algorithm code with line by line:

```python
def knapsack(C, weights, profits):
    """
    Solves the unbounded knapsack problem using bottom-up dynamic programming.
    
    Parameters:
    - C (int): total capacity of the knapsack
    - weights (list of int): weight of each item type
    - profits (list of int): profit of each item type
    
    Returns:
    - dp[C] (int): the maximum profit achievable with full capacity
    - dp (list of int): the full DP table from capacity 0 to C
    """

    # Create a DP table where dp[c] stores the max profit for capacity c
    dp = [0] * (C + 1)  # Initialize all capacities with 0 profit

    # Iterate over every capacity from 1 to C
    for c in range(1, C + 1):
        # Try every item type to see if it can be part of the optimal solution
        for i in range(len(weights)):
            # Only consider the item if it fits in the current capacity
            if weights[i] <= c:
                # Option 1: Don't take this item → keep dp[c] as it is
                # Option 2: Take this item → gain its profit + best from remaining capacity
                dp[c] = max(dp[c], dp[c - weights[i]] + profits[i])
                # We update dp[c] with the better of the two choices

    # dp[C] contains the maximum profit for the full capacity
    return dp[C], dp
```

```python
# Case A: weights = [4,6,8], profits = [7,6,9], C = 14
weights_a = [4, 6, 8]
profits_a = [7, 6, 9]
capacity = 14

max_profit_a, dp_table_a = knapsack(capacity, weights_a, profits_a)

print("Test Case A:")
print("Max Profit =", max_profit_a)
print("DP Table =", dp_table_a)
```
